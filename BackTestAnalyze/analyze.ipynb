```python
import json
```

```python

file_path = 'backtest-result-2025-04-23_11-17-29.json'

with open(file_path, 'r') as file:
    data = json.load(file)

trades = data['strategy']['AwesomeStrategy']['trades']
print(f"Total trades: {len(trades)}")
```

```python
trade = trades[0]
winTrades = [t for t in trades if t['profit_ratio'] > 0]
print(f"Total win trades: {len(winTrades)}")
type (trade)
for key in trade.keys():
    print(key, type(trades[0][key]))
print (trade['open_date'])
print (trade['profit_ratio'])
print(trades[-1]['open_date'])
```

```python
WW2W = 0
WW2L = 0
WL2W = 0
WL2L = 0
LW2W = 0
LW2L = 0
LL2W = 0
LL2L = 0
for i in range(2 , len(trades)):
    if trades[i-2]['profit_ratio'] > 0 and trades[i-1]['profit_ratio'] > 0:
        if trades[i]['profit_ratio'] > 0:
            WW2W += 1
        else:
            WW2L += 1
    elif trades[i-2]['profit_ratio'] > 0 and trades[i-1]['profit_ratio'] < 0:
        if trades[i]['profit_ratio'] > 0:
            WL2W += 1
        else:
            WL2L += 1
    elif trades[i-2]['profit_ratio'] < 0 and trades[i-1]['profit_ratio'] > 0:
        if trades[i]['profit_ratio'] > 0:
            LW2W += 1
        else:
            LW2L += 1
    elif trades[i-2]['profit_ratio'] < 0 and trades[i-1]['profit_ratio'] < 0:
        if trades[i]['profit_ratio'] > 0:
            LL2W += 1
        else:
            LL2L += 1
print (f"WW2W: {WW2W}, WW2L: {WW2L}, WL2W: {WL2W}, WL2L: {WL2L}, LW2W: {LW2W}, LW2L: {LW2L}, LL2W: {LL2W}, LL2L: {LL2L}")
```

```python
print (WW2L + WL2L + LW2L + LL2L + WW2W + WL2W + LW2W + LL2W)
```

```python
print ("proablity of wining next trade after 2 wins: ", WW2W/(WW2W+WW2L))
print ("proablity of wining next trade after 1 win and 1 loss: ", WL2W/(WL2W+WL2L))
print ("proablity of wining next trade after 1 loss and 1 win: ", LW2W/(LW2W+LW2L))
print ("proablity of wining next trade after 2 losses: ", LL2W/(LL2W+LL2L))
```

```python
stopLosses = 0
roies = 0
exitSignal = 0
for trade in trades:
    if (trade['exit_reason'] == 'stop_loss'):
        stopLosses += 1
    elif (trade['exit_reason'] == 'roi'):
        roies += 1
    elif (trade['exit_reason'] == 'exit_signal'):
        exitSignal += 1
print (f"stop losses: {stopLosses}, roies: {roies}, exitSignal: {exitSignal}")
```

```python
# Total trades and wins
total_trades = WW2W + WW2L + WL2W + WL2L + LW2W + LW2L + LL2W + LL2L
total_wins = WW2W + WL2W + LW2W + LL2W

# P(A) - Probability of win
P_A = total_wins / total_trades

print(P_A)
# P(B) - Probability of pattern "WW"
P_B = (WW2W + WW2L) / total_trades
print (P_B)
# P(B | A) - Probability of pattern "WW" given win
P_B_given_A = WW2W / total_wins

# Applying Bayes' Theorem
P_A_given_B = (P_B_given_A * P_A) / P_B

print(f"P(Win | WW) = {P_A_given_B}")
```
